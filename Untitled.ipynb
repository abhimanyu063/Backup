```python
def fiboo(n):
    if n <2:
        return n
    else:
        return fiboo(n-1) + fiboo(n-2)
```

```python
for i in range(1,10):
    print("Fiboo Number --> ", fiboo(i))
```

```python
class Prime:
    def __init__(self, num):
        self.num = num
    def display(self):
        for i in range(self.num):
            if i > 1:
                for j in range(2, i):
                    if i % j == 0:
                        break
                else:
                    print("Prime Number ---> ", i)
```

```python
pr = Prime(100)
pr.display()
```
